```python
def celcius_ke_fahrenheit(celcius):
    return (celcius * 9/5) + 32

print(celcius_ke_fahrenheit(0))
print(celcius_ke_fahrenheit(100))
```

```python
def is_genap(angka):
    if angka % 2 == 0:
        return True
    else:
        return False

print(is_genap(4))
print(is_genap(7))
```

```python
def nilai(angka):
    if angka >= 75:
        return "Lulus"
    else:
        return "Gagal"

print(nilai(80))
print(nilai(60))
```

```python
def bilangan(angka):
    for i in range(1, angka, 2):
        print(i, end=" ")

bilangan(20)
```
